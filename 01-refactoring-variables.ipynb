# **Refactoring Your Code**

**Code refactoring is the process of improving the structure, design, and quality of existing code without changing its functionality**. It involves making changes that make the code easier to understand, modify and maintain in the future. The goal of code refactoring is to improve the code's overall quality and reduce complexity and technical debt.

Refactoring may involve renaming variables or methods to make them more descriptive, simplifying complex code blocks, breaking up large methods or classes into smaller, manageable components, removing duplicated code, and improving overall readability and maintainability.

Code refactoring is an essential practice for software developers, as it helps ensure that the **codebase remains maintainable, adaptable, and scalable**. By continually refactoring, developers can keep the codebase healthy, ultimately **improving the software's quality and performance**.

There are a few things you should consider:

* **Variables**

  * Variable names should reveal intent
  * Use meaningful and pronounceable variable names
  * Use the same vocabulary for the same type of variable
  * Avoid magic numbers and magic strings
  * Use variables to keep code **DRY** (*Don't Repeat Yourself*)
  * Use self-explanatory variables
  * Avoid mental mapping
  * Don’t add unneeded context

* **Dispensable content**

  * Avoid comments (prefer clear code instead)
  * Remove dead code
  * Remove debug-only output before final submission (for example, temporary `print()` calls or one-off exploratory displays)

* **Functions**

  * Use functions to keep code DRY
  * Functions should do one thing
  * Functions should be only one level of abstraction
  * Function names should say what they do
  * Use type hints to improve readability
  * Avoid side effects
  * Avoid unexpected side effects on values passed as parameters
  * Keep function arguments to 2 or fewer ideally
  * Use default arguments instead of short-circuiting or conditionals
  * Don’t use flags as function parameters

### **Exercise 1: Variable names should reveal intent**

We will read more code than we will ever write. It's important for our code to express **intent** so that our readers don't have to waste mental effort to figure out puzzles.

One common culprit in Data Science code is DataFrames. Every DataFrame is named as `df`. In software programming, it's an unusual (and **bad**) practice to embed information about variable types in the variable name (e.g. we would probably never write `string = 'Hello friends'`. Instead, we would write `greeting = 'Hello friends'`).

One rule of thumb on how to name DataFrames is to think about what is in each row. For instance, if each row in my DataFrame is a loan, then the DataFrame is a collection of loan entries. Hence, we could call it `loans`.

Refactor this example so behavior stays the same and readability improves.

````python
def filter_loans(loans_df, month):
    """Return rows for a given month from a DataFrame (supports index or column)."""
    if "month" in loans_df.columns:
        return loans_df[loans_df["month"] == month]
    return loans_df.loc[[month]] if month in loans_df.index else loans_df.iloc[0:0]
````

### **Exercise 5: Use variables to keep code "DRY" ("Don't Repeat Yourself")**

```DRY``` stands for **"Don't Repeat Yourself"**. If you find yourself changing the same thing in multiple places, then this thing is a candidate for refactoring.

Notice how `amount` is repeated in multiple places. If the column name in the data should change (e.g. to `literally_anything`), then we would need to waste effort in finding and replacing `amount` in all these places.

Refactor this example so behavior stays the same and readability improves.

````python
# Bad example
loans = loans.fillna({"amount": 0})
display(loans)
loans.groupby("borrower", as_index=False)["amount"].mean().sort_values(by="amount")
````

````python
# @TODO: Apply DRY by extracting the repeated column name into one variable.
# 1) Define one variable for the target column.
# 2) Reuse it in `fillna`, `groupby` and `sort_values`.
# 3) Keep behavior unchanged.

# Write your refactor below.
````

#### Example Solution

````python
# Difference vs bad example: extract repeated column name to a single variable (DRY).
target_column = "amount"

loans = loans.fillna({target_column: 0})
display(loans)
loans.groupby("borrower", as_index=False)[target_column].mean().sort_values(
    by=target_column
)
````

### **Exercise 6: Avoid mental mapping**

Don’t force the reader of your code to translate what the variable means. **Explicit is better than implicit.**

Refactor this example so behavior stays the same and readability improves.

````python
# Bad example
seq = ("Austin", "New York", "San Francisco")


def do_stuff():
    pass


def do_some_other_stuff():
    pass


def dispatch(value):
    print(f"Dispatching: {value}")


for item in seq:
    do_stuff()
    do_some_other_stuff()
    # ...
    # Wait, what's `item` for again?
    dispatch(item)
````

#### Example Solution

````python
# Difference vs bad example: names are explicit (`locations`, `location`) to avoid mental mapping.
locations = ("Austin", "New York", "San Francisco")


def do_stuff():
    pass


def do_some_other_stuff():
    pass


def dispatch(value):
    print(f"Dispatching: {value}")


for location in locations:
    do_stuff()
    do_some_other_stuff()
    # ...
    dispatch(location)
````

### **Exercise 7: Use explanatory variables**

Refactor this example so behavior stays the same and readability improves.


````python
address = "One Infinite Loop, Cupertino 95014"
````

````python
# Bad example
city_zip_code_regex = r"^[^,\\]+[,\\\s]+(.+?)\s*(\d{5})?$"
address_match = re.match(city_zip_code_regex, address)

if address_match is None:
    raise ValueError("Address format is invalid")

city = address_match.group(1)
zip_code = address_match.group(2)
print(f"City: {city}, Zip Code: {zip_code}")
````

````python
# @TODO: Refactor this regex example with explanatory variables.
# 1) Build a pattern with `?P<city>` and `?P<zip_code>`.
# 2) Match the address.
# 3) Extract and print `city` and `zip_code` using named groups.

# Write your refactor below.
````

#### Example Solution

````python
# Difference vs bad example: named groups make the pattern self-explanatory.
city_zip_code_regex = r"^[^,\\]+[,\\\s]+(?P<city>.+?)\s*(?P<zip_code>\d{5})?$"
address_match = re.match(city_zip_code_regex, address)

if address_match is None:
    raise ValueError("Address format is invalid")

print(
    f"City: {address_match.group('city')}, Zip Code: {address_match.group('zip_code')}"
)
````
