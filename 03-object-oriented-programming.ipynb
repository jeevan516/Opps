# **Object-Oriented Programming**

**Object-oriented programming (OOP)** is a programming paradigm based on the idea of **grouping related data and functions** into “islands” of information called **objects**.

No matter which paradigm is used, programs follow the same basic steps to solve problems:

1. **Data input**: Data is read from a source such as a **file system** or a **database**.  
2. **Processing**: Data is interpreted and possibly altered to prepare it for **display**.  
3. **Data output**: Data is presented so it can be read and interacted with by a **user** or a **system**.



### **Important OOP concepts**  

To reason in OOP, you need to understand fundamental concepts like **classes**, **objects**, and **state**—and know the **differences** and **relationships** between them. You need to know how to work with **variables** and how to add them to your **objects**.



## **Object**

An `object` is like a component of a system. In Python, **an object is an instance of a class**.  

The idea behind OOP is to make programming closer to real-world thinking.

Example analogy:  
- Imagine a **car park**. You see many **cars** of different **shapes**, **sizes**, and **colors**.  
- A car has **properties** like **make**, **model**, and **color**. Assigning values—like a red Ferrari, a four-wheel-drive Jeep, or a yellow Mustang—describes specific objects.

Another example:  
- Picture a **deck of cards**. Each **card** is an object with properties such as **suit** (clubs, hearts, diamonds, spades) and **value** (ace, king, nine, etc.).



## **Class**

A `class` is a data type that acts as a **template or blueprint for a particular kind of object**.

- An OOP system has **objects**, and these objects have **properties**.
- A **class** defines those properties, while the **object** is the actual instance you use.

Example:  
- The **class** is the **blueprint** of a car.  
- The **object** is the **actual car** you drive.

The class is what you write in code, and the object is what is created when the **runtime environment** executes that code with the **variables** you specify.

| **Class** | **Object** |
|-----------|-----------|
| Car       | Honda Accord, Jeep Wrangler |
| Ice Cream | Chocolate, Vanilla, White Coffee |
| Dog       | Snoopy, Pluto, Goofy |

Creating an object from a class is like creating a car from a blueprint:  
- When you create an object, your program requests **resources (memory)** from the operating system.  
- Similarly, when a car is made from a blueprint, the factory requests metal, rubber and glass to build the car.

### **Create an object from a class**

When you create an object from a class, you're said to **instantiate** it. What you are doing is asking the operating system to provide you with enough memory and create an object from the template (the class) with the **initial values** you specify. Basically, **"instantiate"** is another word for **"create"**.

To **instantiate an object**, you add parentheses to the name of the class. What you get is an object that you can choose to **assign to a variable**, like so:

```python
car = Car()
```

`car` is the **variable** that holds your **object's instance**.  
The moment of **instantiation**, or **object creation**, is when you call `Car()`.


## **Variables in OOP vs. Procedural Programming**

From procedural programming, you're used to variables storing information and keeping track of state. You can define these variables wherever you need them in your file.  

The typical **variable initialization** can look like this:


````python
pi = 3.14
````

Variables also exist in OOP, although they are **attached to objects** and are not defined independently.  

**The variables of an object are called attributes**. When an attribute is attached to an object, it's used in one of two ways:

- **Describe the object**: An example of a description variable is the color of a car or the number of spots on a giraffe.  
- **Hold the state**: A variable can instead be used to describe an object's **state**. An example of a state is the floor of an elevator or whether it's running or not.

### **Add attributes to a class**

Knowing what attributes (variables) should be added to your class is part of **modeling**.  

You've learned how to create a class in code, so how do you add an attribute to it? You need to tell the class what attributes it should have at **construction time**, when an object is being instantiated. There's a special function that's called at the moment of creation, called an **initializer**.

#### **Initializer**

In many programming languages there is the notion of an **initializer**, a special function that's called only when an object is first created. **The initializer is called only once**.  

In this method, you create the attributes that the object should have. Additionally, you assign any **initial values** to the created attributes.

In Python, the initializer has the name **`__init__()`**.  

You also need to pass a special parameter, **`self`**, to the initializer.  

The parameter `self` refers to the object's instance. Any assignment to this keyword means that the attribute ends up on the object instance. If you don't add an attribute to `self`, it will instead be treated as a **temporary variable** that won't exist after `__init__()` is done executing.

> **Note:** The parameter `self` must also be passed to any **methods** that need to refer to anything on the object instance.  
> This concept will be covered in the next unit.

#### **Add and initialize attributes on a class**

Let's see an example of setting up attributes in an initializer:


````python
class Car:  # Define a class named Car
    def __init__(self, color):  # Define the constructor method for the Car class
        self.make = "Mercedes"  # Set the make attribute to "Mercedes"
        self.color = color  # Set the color attribute to the value passed as an argument


car = Car("black")  # Create an instance of the Car class with the color "black"

print(car.make)
print(car.color)
````

The preceding example describes the class `Car` with two variables, `make` and `color`.  

An important takeaway from the code is that `__init__()` **is called implicitly**. You don't call the `__init__()` method by name, but it's called when the object is created, in this line of code:

```python
car = Car('black')
```

## **Methods in OOP**

No matter what paradigm is used, **methods can perform an action**. That action can be a computation that relies only on inputs, or it can change the value of a variable.

Methods on objects in OOP come in two flavors:

- **External methods**, which other objects can invoke.  
- **Internal methods**, which cannot be reached by other objects. Additionally, internal methods help to execute a task that was started by an invocation to an external method.

No matter what type of method, they can change the value of an object's attribute; in other words, its **state**.

The notion of state, and who and what can change it, is an important part of designing your classes and objects.  

This leads us into our next section, **encapsulation**.


## **Encapsulation: Protect your data**

The general idea of **encapsulation** is that the **data on an object is internal**, something that only concerns the object. The data is needed for the object and methods to do their job, which is performing a specific task. When you say that the data is **internal**, it means that it should be **protected from other manipulation from the outside**, or rather from uncontrolled manipulation from the outside.  

The question is: **why?**

### **Why you need it**

Let's explain the reasoning why data shouldn't be directly touched by another object.  

Here are some examples:

- **You don't need to know the internals.**  
  When you drive a car, you interact with it using simple controls like the pedals or the steering wheel. You don’t need to understand how the engine, transmission or braking system works internally, you just use the interface provided to achieve the desired result.

  Programming works the same way. When using an object, you don’t need to know how it performs a task internally. You simply call its methods and trust that they do what they’re supposed to do.

- **You shouldn't know the internals.**  
  A driver is meant to use the steering wheel and pedals, not open the hood and manipulate wires or engine parts while driving. Hiding those details makes the car safer and more reliable.  

  In the same way, an object should expose a clear public interface and keep its internal state protected. If other parts of the program can directly modify that internal state, they can accidentally put the object in an invalid or inconsistent state.

Let's have a look at a more concrete example, a `Square` class with the following code:


````python
class Square:
    def __init__(self):
        self.height = 2
        self.width = 2

    def set_side(self, new_side):
        self.height = new_side
        self.width = new_side


square = Square()
print(square.height)
square.height = 3  # Not a square anymore
print(square.height)
````

````python
square.set_side(3)
print(square.width)
````

In the `Square` example, you break the notion of what a square is by setting the height variable. The way the square is coded, it needs you to invoke the method `set_side()` for the square to function properly.  

**Letting the object take care of its data is considered safer**. In nearly every instance, you should **choose to interact via a method** versus setting the data explicitly.

### **Access levels**

How can you protect your class and your object from unwanted manipulation of data? The answer is with **access levels**.  

You can **hide data** from the outside world, and from other objects, by marking data and functions with specific keywords. These keywords are known as **access modifiers**.

The way Python accomplishes data hiding is by **adding prefixes to attribute names**. One leading underscore **`_`** is a message to the outside world that this data probably shouldn't be touched.  

When you modify the `Square` class, you end up with this code:


````python
class Square:
    def __init__(self):
        self._height = 2
        self._width = 2

    def set_side(self, new_side):
        self._height = new_side
        self._width = new_side


square = Square()
print(square._height)
square._height = 3  # Not a square anymore
print(square._height)
````

**One** leading underscore still allows the attribute to be modified and is only a convention that signals "internal use".  

Can we do this better? You can use **two** leading underscores **`__`** to trigger **name mangling**, which makes accidental access harder.  

Your `Square` class should now look like this:


````python
class Square:
    def __init__(self):
        self.__height = 2
        self.__width = 2

    def set_side(self, new_side):
        self.__height = new_side
        self.__width = new_side


square = Square()
````

````python
try:
    print(getattr(square, "__height"))
except AttributeError as error:
    print(f"Expected AttributeError: {error}")
````

Great, so we're safe.  

Have we protected our data? Well, not entirely. Python just changes the name of the underlying data.  

By entering this code, you can still change its value:


````python
square = Square()
mangled_height = "_Square__height"
print(getattr(square, mangled_height))  # Is allowed
````

Many other languages implement stricter access control than Python. In Python, privacy is mostly based on conventions and name mangling, not hard runtime enforcement.

## **Getters and setters**

We've said so far that **data in general shouldn't be touched from the outside**. Data is the object's concern.  

As with all rules and strong recommendations, there are **exceptions**. Sometimes you need to change the data, or changing it is simpler than having to add a significant amount of code.

**Getters** and **setters**, which are also known as **accessors** and **mutators**, are methods dedicated to reading or changing your data.  

- **Getters** expose read access to internal state in a controlled way.  
- **Setters** update internal state and can enforce validation rules. They act as a guard so that a bad value can't be set.  

Let's bring up our `Square` class again and see getters and setters in action:


````python
class Square:
    def __init__(self):
        self.__height = 2
        self.__width = 2

    def set_side(self, new_side):
        self.__height = new_side
        self.__width = new_side

    def get_height(self):
        return self.__height

    def set_height(self, h):
        if h >= 0:
            self.__height = h
        else:
            raise ValueError("Value needs to be 0 or larger")


square = Square()
print(square.get_height())
square.set_height(3)
print(square.get_height())

try:
    square.set_height(-3)
except ValueError as error:
    print(f"Expected ValueError: {error}")
````

The method `set_height()` protects you from setting the value to something negative.  If you do, it raises an exception.

### **Use decorators for getters and setters**

**Decorators** are an important subject in Python. They're part of a larger subject called **meta-programming**. A decorator **wraps or transforms function behavior in a reusable way**.

In the context of OOP and getters/setters, the `@property` pattern reduces boilerplate:

- `@property` marks a method as a **getter**.
- `@<name>.setter` marks a method as the corresponding **setter**.
- You can include validation logic in the setter before saving values.

> Note: `@property` does **not** automatically create storage for you. You still define the backing attribute (for example, `self.__height`) yourself.

Let's see this **decorator** in action:


````python
class Square:
    def __init__(self, h, w):
        self.__height = h
        self.__width = w

    def set_side(self, new_side):
        self.__height = new_side
        self.__width = new_side

    @property
    def height(self):
        return self.__height

    @height.setter
    def height(self, new_value):
        if new_value >= 0:
            self.__height = new_value
        else:
            raise ValueError("Value must be greater than or equal to 0")


square = Square(2, 2)
print(square.height)
square.height = 3
print(square.height)

try:
    square.height = -3
except ValueError as error:
    print(f"Expected ValueError: {error}")
````

In the preceding code, `self.__height` is the **private field** (more precisely: a name-mangled attribute). It is created in `__init__` with `self.__height = h`, **not** by `@property`.  

The `@property` decorator creates a **getter interface** so you can read that internal value via `square.height` instead of direct access to `square.__height`.  

The `@height.setter` decorator defines how assignments like `square.height = 3` are handled. In other words, it routes writes through validation logic before updating `self.__height`.  

One caveat remains: if your class models a square, changing only `height` can break the `height == width` invariant. Before exposing getters/setters, decide which state changes are allowed and enforce those rules in your methods/properties.


## **Inheritance**

In Python, you can get the features you want from an existing class (**parent**) to create a new class (**child**). This Python feature is called **inheritance**.

By **inheritance**, you can:

- obtain the features of a parent class,  
- change the features that you don’t need,  
- add new features to your child class (**derived class** or **subclass**).

Since you are using a pre-used, tested class, you don’t have to put quite as much effort into your new class. **The child class inherits the attributes and functions of its parent class**.

If we have several similar classes, we can define the **common functionalities** of them in one class, define child classes of this parent class and **implement specific functionalities** there.

We have 3 data professionals. They all know mathematics and statistics, a programming language like Python, R or Java, and reached some score in an SQL test.

````python
class Data_Scientist:
    def __init__(self, name, sql_test_score):
        self.name = name
        self.sql_test_score = sql_test_score

    def knows_maths_stats(self):
        return True

    def knows_programming(self):
        return True


class Data_Analyst:
    def __init__(self, name, sql_test_score):
        self.name = name
        self.sql_test_score = sql_test_score

    def knows_maths_stats(self):
        return True

    def knows_programming(self):
        return True


class Data_Engineer:
    def __init__(self, name, sql_test_score):
        self.name = name
        self.sql_test_score = sql_test_score

    def knows_maths_stats(self):
        return True

    def knows_programming(self):
        return True


example_scientist = Data_Scientist("Example Scientist", 8)
example_analyst = Data_Analyst("Example Analyst", 7)
example_engineer = Data_Engineer("Example Engineer", 9)

print(
    example_scientist.knows_programming(),
    example_analyst.knows_programming(),
    example_engineer.knows_programming(),
)
````

Instead of writing the same class again and again, we can define a parent class `Data_Professional` and 3 child classes of this class: `Data_Scientist`, `Data_Analyst` and `Data_Engineer`.

### **Parent Class**

The class from which we inherit is called the **parent class, superclass or base class**.


````python
class Data_Professional:
    def __init__(self, name, sql_test_score):
        self.name = name
        self.sql_test_score = sql_test_score

    def knows_maths_stats(self):
        return True

    def knows_programming(self):
        return True
````

### **Child Classes**

The new class which was created by inheriting functionalities of the parent class is called **child class, derived class or subclass**. The child class’s `__init__()` function **overrides** the parent class’s `__init__()` function.

#### **Accessing parent class variables from child class**

Mary Smith is a data scientist who knows maths and stats, SQL, and Python.

Define the child class:


````python
class Data_Scientist(Data_Professional):
    def __init__(self, name, sql_test_score):
        super().__init__(name, sql_test_score)
````

Initialize `dt_sci` object and the features of `dt_sci`:


````python
dt_sci = Data_Scientist("Mary Smith", 9)

print(
    dt_sci.name,
    dt_sci.sql_test_score,
    dt_sci.knows_maths_stats(),
    dt_sci.knows_programming(),
)
````

An empty class has the properties of its parent class:


````python
class Data_Engineer(Data_Professional):
    pass


dt_en = Data_Engineer("Hyun-mi Tokko", 9)
print(
    dt_en.name,
    dt_en.sql_test_score,
    dt_en.knows_maths_stats(),
    dt_en.knows_programming(),
)
````

Overriding methods from parent classes:


````python
class Data_Analyst(Data_Professional):
    def __init__(self, name, sql_test_score):
        super().__init__(name, sql_test_score)

    def knows_programming(self):
        return False


dt_an = Data_Analyst("Brinda Kumar", 7)

print(
    dt_an.name,
    dt_an.sql_test_score,
    dt_an.knows_maths_stats(),
    dt_an.knows_programming(),
)
````

### **Advantages of inheritance**

Inheritance provides a lot of advantages. By using parent and child classes we can:

- **Reduce code duplication** by putting common attributes and methods into the parent class.
- **Improve code organization and readability**, making it easier to debug and maintain.
- **Create flexible, reusable code** that's easy for developers to understand and extend.

#### Example Solution

````python
# Difference vs bad example: account behavior and validation are encapsulated in classes.
class BankAccount:
    def __init__(self, account_holder: str, balance: float):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount: float) -> None:
        self.balance += amount

    def withdraw(self, amount: float) -> None:
        if amount > self.balance:
            raise ValueError("Insufficient funds")
        self.balance -= amount

    def get_balance(self) -> float:
        return self.balance


class CheckingAccount(BankAccount):
    TRANSACTION_FEE = 1.50

    def withdraw(self, amount: float) -> None:
        total = amount + self.TRANSACTION_FEE
        if total > self.balance:
            raise ValueError("Insufficient funds")
        self.balance -= total


account = CheckingAccount("Alex", 100.0)
account.deposit(25.0)
account.withdraw(40.0)
print(account.get_balance())

try:
    account.withdraw(1000.0)
except ValueError as error:
    print(f"Expected ValueError: {error}")
````
